## Imports 

```python
import sqlite3
import pandas as pd
```

## Connect with sqlite db

```python
connect = sqlite3.connect("../data/database.sqlite")
```

## Load data

#### Read and List all tables from SQLite db

```python
tables = pd.read_sql("SELECT NAME from sqlite_master WHERE type='table';", connect)[
    "name"
].tolist()
```

```python
tables
```

#### dump each table into a dict of dataframes


```python
dfs={t:pd.read_sql(f"SELECT  * from {t}", connect) for t in tables}
```

```python
type(dfs)
```

```python
type(dfs.items())
```

## Inspect Datasets

```python
for name, df in dfs.items():
    print(f"\nName: {name}, Shape: {df.shape}")
    print("\nData Types of every column: ")
    print(df.dtypes) # numeric or str col
    print("\nCounting nulls :")
    print(df.isnull().sum())
    print()
```

# Check Missing Values

```python
print("Missing Values: ")
for t in table_names:
    print("Name: ", t)
    print(dfs[f'{t}'].isnull().sum())
```

### Cols with missing values

```python
print("Missing values per column:")
for t in table_names:
    print(f"Name: {t}")
    missing=dfs[f'{t}'].isnull().sum().sort_values(ascending=False)
    missing=missing[missing>0]
    print(missing)
```
